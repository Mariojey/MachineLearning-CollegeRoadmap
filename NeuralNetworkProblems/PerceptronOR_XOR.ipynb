```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
X_OR = torch.tensor([
    [0.,0.],
    [0.,1.],
    [1.,0.],
    [1.,1.]
])

Y_OR = torch.tensor([
    [0.],
    [1.],
    [1.],
    [1.]
])

X_XOR = torch.tensor([
    [0.,0.],
    [0.,1.],
    [1.,0.],
    [1.,1.]
])

Y_XOR = torch.tensor([
    [0.],
    [1.],
    [1.],
    [0.]
])
```

z = w_{1} * x_{1} + w_{2} * x_{2} + b
Two inputs and one output neuron

Activation function - Sigmoid:

sigma(x) = 1/(1 + e^(-x))

This eq changes domain to (0;1)

```python
class Perceptron(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(2,1)
        self.activation = nn.Sigmoid()

    def forward(self, x):
        x = self.linear(x)
        x = self.activation(x)
        return x
```

lr in optim's SGD means learning rate

Binary Cross Entropy for loss function:

L = -(ylog(p) + (1-y)log(1-p))

y - real value
p - model's prediction

```python
def train_model(X, y, epochs=5000):

    model = Perceptron()

    criterion = nn.BCELoss()
    optimizer = optim.SGD(model.parameters(), lr=0.1)

    for epoch in range(epochs):
        y_pred = model(X)

        loss = criterion(y_pred, y)

        optimizer.zero_grad()

        loss.backward()
        optimizer.step()

        if epoch % 1000 == 0:
            print(f"Epoch {epoch} Loss {loss.item():.4f}")

    return model
```

SGD changes weights in nn to minimalize error.

```python
print("TRAIN ON")
model_OR = train_model(X_OR, Y_OR)
```

```python
with torch.no_grad():
    pred = model_OR(X_OR)
    pred_class = (pred > 0.5).float()
```

```python
print("Prediction for OR")
print(pred_class)
```

And that's what we want

```python
print("TRAIN XOR")
model_XOR = train_model(X_XOR, Y_XOR)
```

```python
with torch.no_grad():
    pred = model_XOR(X_XOR)
    pred_class = (pred > 0.5).float()
```

```python
print("Prediction for XOR:")
print(pred_class)
```
